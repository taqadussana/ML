## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** Which pages should a content editor spend their limited review time on first?

**Unit of analysis:** one content page (`content_id`).

**Who acts on it:** a content editor or SEO reviewer with a fixed weekly review capacity —
they can't check every page, so they need a short, trustworthy shortlist.

**The action:** the editor opens the top-ranked pages, checks the reason code attached to each
one, and decides whether to refresh, expand, consolidate, or leave it alone.

**Cost of a wrong call:** two different failure directions, both with a real cost. If the model
ranks a healthy page too high, the editor burns limited review time on a page that didn't need
it — that's wasted capacity. If it misses a page that's genuinely declining, that page keeps
losing visibility unnoticed until someone catches it by chance — that's a missed opportunity
that compounds the longer it goes unreviewed. Because review capacity is scarce, I care more
about precision in the top of the ranking than about catching every single case.

**Why a plain rule isn't enough:** a single if/else like "stale AND visible" catches the obvious
cases, but real decline shows up as a tangle of weaker signals — position, CTR, freshness,
engagement — that don't cleanly separate with one or two conditions. That's exactly the kind of
messy-but-real pattern a model earns its place on.
